1. Write a program that asks the user for a number of days. The program then
prints out the number of seconds in the amount of days given

````python
print ("This calculator calculates seconds given the number of days a user inputs")

def findseconds(my_userinput):
    seconds_in_a_minute=60 # 1 min= 60 seconds
    seconds_in_an_hour= 60 * seconds_in_a_minute #1 hour = 60 minutes seconds
    seconds_in_a_day = 24 * seconds_in_an_hour # 1 day = 24 hours
    return my_userinput*seconds_in_a_day
````

````python
user_input_as_number_of_days=int(input("Enter Number of days: "))
````

````python
print(findseconds(user_input_as_number_of_days))
````

2. Write a program that asks a user to input the radius then the program
calculates the volume of a sphere (the formula for the volume is (4/3)πr3
).
Use the exponential operator in python to compute (r3
).

````python
import math

#I am importing the Maths library
#that utilizes the pow function that takes in base and exponent.
#i'm also using the attributed pi value
#In our case the base is the radius and the exponent is 3
````

````python

def findspherevolume(userinput):
    volume=(4/3) * math.pi * math.pow(userinput,3)
    return volume
````

````python
radius_prompt=float(input("Enter Radius of the sphere: "))

print ("The volume of the sphere is: ",findspherevolume(radius_prompt))
````

3. Using functions, write a program to compute the area and perimeter of a
square. The program should ask the user to enter a number corresponding to
the side length of the square and display the area and perimeter of the
square

````python
def find_square_perimeter(side):
    perimeter=side*4
    return perimeter

def find_square_area(side):
    return math.pow(side,2)

    
````

````python
userinput=int(input("Enter the side length of the square: "))
print (f"Perimeter: {find_square_perimeter(userinput)}\nArea: {find_square_area(userinput)}")
````

4.Write a program using functions that determines whether a character input
by a user is uppercase or lower case.

````python

def checkLetter(mycharacter):
    if mycharacter.isupper():
        return f"{mycharacter} is an uppercase letter"
    elif mycharacter.islower():
        return f"{mycharacter} is a lowercase letter"
    else:
        return "Not a letter"

characterinput=str(input("Enter a character: "))
````

````python
print (checkLetter(characterinput))
````

6. Write a Python program that does the following:

- Uses a loop for a user to continually input 5 values to populate an
array.
- Calculates and displays the average of the values input into the array.

````python
values=[]
count=0
while (count<5):
    value=int(input(f"Enter value {count+1}: "))
    # Every time a user inputs a value, we append it
    # Until the required limit (5) is reached
    values.append(value)
    count+=1
````

````python
print(values)
````

````python
def find_average(values):
    accumulated_sum=0
    for i in range(len(values)):
        accumulated_sum+=values[i]
    avg=accumulated_sum/len(values)
    return avg

print(find_average(values))
````

````python
from typing import List
class solution:
    def absDistinctValues(self, numbers: List[int])->int:
        counter=1 #There should be at least 1 absolute distinct value
        head = 0
        tail = len(numbers) -1
        my_max_value=max(abs(numbers[head]), abs(numbers[tail]))

        while head<=tail:
            
            head_val=abs(numbers[head])
            if head_val==my_max_value:
                head+=1
                continue
            tail_val=abs(numbers[tail])
            if tail_val== my_max_value:
                tail-=1
                continue
            if head_val>=tail_val:
                my_max_value=head_val
                head+=1
            else:
                my_max_value=tail_val
                tail-=1
            counter+=1

        return counter
                
            
            
        
````
